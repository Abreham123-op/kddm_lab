## 2. Load the source tables into a **raw schema**

A Schema is a namespace that groups tables. We keep the source copies in **raw**
so that they remain distinguishable from cleaned warehouse tables we will create later.

### Define a mapping to transform column types

````python
RAW_TABLES: dict[str, dict[str, str]] = {
    "customers": {"customer_zip_code_prefix": "VARCHAR"},  # keep leading zeros
    "orders": {},
    "order_items": {},
    "products": {},
    "sellers": {}
}
````

### Define some helper functions

````python
def csv_path(table: str) -> Path:
    return RAW_DIR / f"olist_{table}_dataset.csv"


def check_source_files(tables) -> None:
    missing = [csv_path(table) for table in tables if not csv_path(table).is_file()]
    if missing:
        names = "\n".join(f" - {path}" for path in missing)
        raise FileNotFoundError(f"Missing CSV files:\n{names}")
````

#### Define schema name for source data

````python
RAW_SCHEMA = "raw"
````

### Load data into source tables

````python
def load_raw_table(con: duckdb.DuckDBPyConnection, table: str, column_types: dict[str, str]) -> int:
    target = f'{RAW_SCHEMA}."{table}"'
    source = con.read_csv(str(csv_path(table)), header=True, dtype=column_types or None)
    con.execute(f"DROP TABLE IF EXISTS {target}")
    source.to_table(target)
    return con.execute(f"SELECT count(*) FROM {target}").fetchone()[0]
````

### Sanity check if we have all required source files

````python
check_source_files(RAW_TABLES)
````

### Create **RAW_SCHEMA** if it does not exist

````python
con.execute(f"CREATE SCHEMA IF NOT EXISTS {RAW_SCHEMA}")
````

### Execute **load_raw_table** to load all source data

````python
row_counts = {t: load_raw_table(con, t, types) for t, types in RAW_TABLES.items()}
````

### Check count of rows inserted into each `raw`.`<table>`

````python
pd.DataFrame(row_counts.items(), columns=["table_name", "row_count"])
````

## 4. Primary Keys

A **primary key** uniquely identifies a row. A column is a valid candidate only if `count(*) = count(distinct ...)` and it has no nulls.

- A **composite key** uses several columns together (`order_items`)
- A **natural key** comes from the business (`customer_unique_id`).
- A **surrogate key** is generated by the warehouse (`customer_key = 1, 2, 3...`)

````python
q("""
SELECT 'customers.customer_id' AS candidate, count(*) AS n_rows,
       count(DISTINCT customer_id) AS n_distinct,
       count(*) FILTER (WHERE customer_id IS NULL) AS null_rows
FROM raw.customers
UNION ALL
SELECT 'customers.customer_unique_id', count(*), count(DISTINCT customer_unique_id),
       count(*) FILTER (WHERE customer_unique_id IS NULL)
FROM raw.customers
UNION ALL
SELECT 'orders.order_id', count(*), count(DISTINCT order_id),
       count(*) FILTER (WHERE order_id IS NULL)
FROM raw.orders
UNION ALL
SELECT 'order_items(order_id, order_item_id)', count(*),
       count(DISTINCT (order_id, order_item_id)),
       count(*) FILTER (WHERE order_id IS NULL OR order_item_id IS NULL)
FROM raw.order_items
UNION ALL
SELECT 'products.product_id', count(*), count(DISTINCT product_id),
       count(*) FILTER (WHERE product_id IS NULL)
FROM raw.products
UNION ALL
SELECT 'sellers.seller_id', count(*), count(DISTINCT seller_id),
       count(*) FILTER (WHERE seller_id IS NULL)
FROM raw.sellers
ORDER BY candidate
""")
````

## 4. Foreign key relationships

```
customers 1 ---- N orders 1 ---- N order_items N ---- 1 products
```

A foreign key is healthy when every value exists in the parent table. An **orphan** is
a child row with no parent.

````python
q("""
SELECT 'orders without customer' AS check_name, count(*) AS n
  FROM raw.orders o ANTI JOIN raw.customers c USING (customer_id)
UNION ALL SELECT 'items without order', count(*)
  FROM raw.order_items i ANTI JOIN raw.orders o USING (order_id)
UNION ALL SELECT 'items without product', count(*)
  FROM raw.order_items i ANTI JOIN raw.products p USING (product_id)
UNION ALL SELECT 'items without seller', count(*)
    FROM raw.order_items i ANTI JOIN raw.sellers s USING (seller_id)
UNION ALL SELECT 'orders without items', count(*)
  FROM raw.orders o ANTI JOIN raw.order_items i USING (order_id)
""")
````

No orphans in the child-to-parent direction. But 775 orders have no items. An **INNER JOIN** from orders
to items would silently remove them.

## 5. Data Quality

we check five dimensions: completeness(nulls), uniqueness, validity (values in a sensible range),
consistency (fields agree), timeliness (coverage)

### 5.1 Orders with no items

#### Exercise 1

````python
q("""
SELECT count(*) AS orders_without_items
FROM raw.orders AS o
ANTI JOIN raw.order_items AS i USING (order_id)
""")
````

### 5.2 Check order status and class imbalance

#### Exercise 2

````python
q("""
SELECT
    order_status,
    count(*) AS order_count,
    round(100.0 * count(*) / sum(count(*)) OVER (), 2) AS percent_of_orders
FROM raw.orders
GROUP BY order_status
ORDER BY order_count DESC
""")
````

### 5.3 Missing and inconsistent timestamps

#### Exercise 3

````python
q("""
SELECT
    count(*) FILTER (WHERE order_purchase_timestamp IS NULL) AS missing_purchase_timestamp,
    count(*) FILTER (WHERE order_approved_at IS NULL) AS missing_approved_timestamp,
    count(*) FILTER (WHERE order_delivered_carrier_date IS NULL) AS missing_carrier_timestamp,
    count(*) FILTER (WHERE order_delivered_customer_date IS NULL) AS missing_customer_delivery_timestamp,
    count(*) FILTER (WHERE order_estimated_delivery_date IS NULL) AS missing_estimated_delivery_timestamp,
    count(*) FILTER (
        WHERE order_approved_at < order_purchase_timestamp
    ) AS approved_before_purchase,
    count(*) FILTER (
        WHERE order_delivered_carrier_date < order_approved_at
    ) AS carrier_before_approval,
    count(*) FILTER (
        WHERE order_delivered_customer_date < order_delivered_carrier_date
    ) AS delivered_before_carrier,
    count(*) FILTER (
        WHERE order_delivered_customer_date > order_estimated_delivery_date
    ) AS delivered_after_estimated_date
FROM raw.orders
""")
````

### 5.4 Time coverage

#### Exercise 4

````python
monthly = q("""
SELECT
    strftime(order_purchase_timestamp, '%Y-%m') AS month,
    count(*) AS orders
FROM raw.orders
WHERE order_purchase_timestamp IS NOT NULL
GROUP BY month
ORDER BY month
""")
````

### 5.5 Product completeness check

#### Exercise 5

````python
q("""
SELECT
    count(*) AS total_products,
    count(*) FILTER (WHERE product_category_name IS NULL) AS missing_category,
    count(*) FILTER (WHERE product_name_lenght IS NULL) AS missing_name_length,
    count(*) FILTER (WHERE product_description_lenght IS NULL) AS missing_description_length,
    count(*) FILTER (WHERE product_photos_qty IS NULL) AS missing_photo_count,
    count(*) FILTER (WHERE product_weight_g IS NULL) AS missing_weight,
    count(*) FILTER (WHERE product_length_cm IS NULL) AS missing_length,
    count(*) FILTER (WHERE product_height_cm IS NULL) AS missing_height,
    count(*) FILTER (WHERE product_width_cm IS NULL) AS missing_width
FROM raw.products
""")
````

## 6. Relationship cardinality

This query counts child rows for every parent, including parents with zero children, to measure the relationship ranges in the supplied dataset.

````python
q("""
WITH customer_orders AS (
    SELECT c.customer_id, count(o.order_id) AS child_rows
    FROM raw.customers AS c
    LEFT JOIN raw.orders AS o USING (customer_id)
    GROUP BY c.customer_id
), order_items AS (
    SELECT o.order_id, count(i.order_id) AS child_rows
    FROM raw.orders AS o
    LEFT JOIN raw.order_items AS i USING (order_id)
    GROUP BY o.order_id
), product_items AS (
    SELECT p.product_id, count(i.product_id) AS child_rows
    FROM raw.products AS p
    LEFT JOIN raw.order_items AS i USING (product_id)
    GROUP BY p.product_id
), seller_items AS (
    SELECT s.seller_id, count(i.seller_id) AS child_rows
    FROM raw.sellers AS s
    LEFT JOIN raw.order_items AS i USING (seller_id)
    GROUP BY s.seller_id
)
SELECT 'customers -> orders' AS relationship, min(child_rows) AS min_children,
       max(child_rows) AS max_children FROM customer_orders
UNION ALL
SELECT 'orders -> order_items', min(child_rows), max(child_rows) FROM order_items
UNION ALL
SELECT 'products -> order_items', min(child_rows), max(child_rows) FROM product_items
UNION ALL
SELECT 'sellers -> order_items', min(child_rows), max(child_rows) FROM seller_items
ORDER BY relationship
""")
````
